```python
#shifting dates with offsets
from pandas.tseries.offsets import Day, MonthEnd
now = datetime(2016, 11, 14)
```

```python
#shift 3 days
now + 3*Day()
```

```python
#shifting to the end of the month
now + MonthEnd()
```

```python
#shifting to the end of next month
now + MonthEnd(2)
```

```python
offset = MonthEnd()
offset.rollforward(now)
```

```python
offset.rollback(now)
```

```python
#create a time series with B = BusinessDay frequency
rng = pd.date_range('3/7/2012 9:30', periods=10, freq='B')
ts = pd.Series(np.random.randn(len(rng)), index=rng)
ts
```

```python
#create some time zone aware timestamps
ts1 = ts[:7].tz_localize('Europe/London')
ts1
```

```python
#convert some timestamps
ts2 = ts1[2:].tz_convert('Europe/Budapest')
ts2
```

```python
#combine result - the result will be UTC
result = ts1 + ts2
result.index
```

```python
#creating a series
rng = pd.date_range('2000-01-01', periods=12, freq='ME')
ts = pd.Series(np.random.randn(12), index=rng)
ts
```

```python
#create a time series and resample
rng = pd.date_range('2000-01-01', periods=365, freq='D')
ts = pd.Series(np.random.randn(len(rng)), index=rng)
ts.head()
```

```python
#resample using mean
ts.resample('ME').mean()
```

```python
#resample by period using mean
ts.resample('ME', kind='period').mean()
```

```python
#downsampling
#creating a time series with 12 minutes
rng = pd.date_range('2000-01-01', periods=12, freq='min')
ts = pd.Series(np.arange(12), index=rng)
ts
```

```python
#aggregate the data into 5min chunks
ts.resample('5min', closed='right').sum()
```
