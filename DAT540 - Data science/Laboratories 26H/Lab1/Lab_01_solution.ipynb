### Question 6: Example of for loop and if condition inside a function

```python
# mylist = [int1, int2, int3]
# create the list 'mylist' with 3 integers in order to print only one of these integers

def filter(seq):
    for x in seq:
        if 0 <= x <= 100:
            continue
        print(x)
```

```python
# ANSWER

mylist = [1, 20, 102]
filter(mylist)

# Notes: 
# 1) only argument 102 is outside [0, 100]
# 2) 'continue' permits to continue the if condition for all values of the list 'seq'
```

### Question 8: Example of function to call another function and to change the type of a variable

```python
# What is the final values of the variables a and b ?
# Try to understand the intermediate changes of these variables
a = 3
b = 4

def changetype(x):
    b = square(x)
    a = multiplication()
    return str(a), str(b)
```

```python
# ANSWER

a = 3
b = 4

def changetype(x):
    b = square(x) # square(3) = 3^2 = 9 is assigned to the local variable b
    print(b) 
    a = multiplication() # multiplication of 2 global variables a b = 3 x 4 = 12 assigned to the local variable a  
    print(a)
    return str(a), str(b) # the 2 local variables are returned as string type

changetype(3)

# Note: pay attention between local and global variables
```
