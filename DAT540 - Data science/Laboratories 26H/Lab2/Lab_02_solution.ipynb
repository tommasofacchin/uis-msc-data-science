### Question 2
What is the difference between global and local variables ?

Create two functions to understand the difference between a local and a global variable.

```python
x = 10

def func_global(y):
  global x  # the variable x is declared globally (out of the function)
  print('x + y =', x + y)

func_global(2)


def func_local(y):
  x = 2  # the variable x is declared locally
  print('x + y =', x + y)

func_local(2)
```

#### NOTE:
* func_global(y) and func_local(y) are using the variable x

* *global variable* x:
    * the global variable x is defined *outside* of the function (x=10 and func_global(y))
    * the global variable x is requested as `global` x
    * the global variable may change inside the function (see func_global2(y))

* *local variable* x:
    * the local variable x is defined *inside* the function
    * if you initialize a variable x and then ask the same object (variable x) as global variable already defined ouside the function => `SyntaxError` (see func_oups(y))

```python
def func_global2(y):
  global x  # the variable x is declared globally (outside the function)
  x = 2
  print('x + y =', x + y)

func_global2(2)
```

```python
def func_oups(y):
  x = 2
  global x  # the variable x is declared globally (out of the function)
  print('x + y =', x + y)

func_oups(2)
```

Create a function with one attribute (guess):

* prints "r = (value of r), too big !" if your guess is larger than r
* prints "r = (value of r), too small !" if your guess is smaller than r
* prints "r = (value of r), correct !" if your guess is equal to r

*Expected if random integer r = 4*:

* `r = 4 , too high !` OR

```python
# ANSWER
import random

def guess_number(guess):
    """
    Generates a random value between 1 and 10 both included.
    Prints whether the number guessed by the user is higher, lower or equal to the random number.

    Args:
    guess (integer)

    Prints:
    "too big !" if the guess is too big
    "too small !" if the guess is too small
    "correct !" if the guess is correct
    """
    r = random.randint(1, 10)
    if guess > r:
        print("r =", r, ", too high !")
    elif guess < r:
        print("r =", r, ", too low !")
    else:
        print("r =", r, ", correct !")
```

### Question 6
Create a function with 2 attributes (length, width) that return a tuple (area, perimeter) with:
* the area of the rectangle calculated with the 2 attributes
* the perimeter of the rectangle calculated with the 2 attributes

*Expected if length = 10 and width = 2.5*: `(25.0, 25.0)`

```python
# ANSWER
def compute_rectangle_area_perimeter(length, width):
    """
    Computes the area and perimeter of a rectangle

    Args:
    length: the length of the rectangle
    width: the width of the rectangle

    Returns:
    a tuple of (area, perimeter)

    Raises:
    an AssertionError if inputs are negative.
    """
    # AssertionError if any of the inputs is negative.
    assert length >= 0 and width >= 0, "Length and width must be non-negative"

    area = length * width
    perimeter = (length + width) * 2
    return area, perimeter

# NOTE: It could be interesting to get a specific error message (AssertionError) if one argument is not correct
```

### Question 8
Create a function - using the function divmod()- that convert a number of seconds in a tuple (days, hours, minutes, seconds)

*Expected if the number of seconds is 129000*:

`(1, 11, 50, 0)`

*Hint*: the function `divmod()` returns the quotient and the remainder in a single call.

```python
# ANSWER
def convert_seconds(num_seconds):
    """
    Converts seconds to days, hours, minutes and seconds.

    Args:
    num_seconds: number of seconds.

    Returns:
    A 4-tuple of integers that represent the number of days, hours, minutes and seconds.

    Raises:
    An AssertionError if the input is negative.
    """
    # AssertionError if any of the inputs is negative.
    assert num_seconds >= 0, "The number of seconds must be non-negative"

    num_minutes, num_seconds = divmod(num_seconds, 60)
    num_hours, num_minutes = divmod(num_minutes, 60)
    num_days, num_hours = divmod(num_hours, 24)
    return num_days, num_hours, num_minutes, num_seconds
```

```python
convert_seconds(129000)
```

```python
convert_seconds(-129000)
```

### Question 10
Create a function that computes the total number of seconds from days, hours, minutes and seconds declared as arguments

*Expected result if (num_days, num_hours, num_minutes, num_seconds) = (1,2,3,4)*: `93784`

```python
# ANSWER
def compute_seconds(num_days, num_hours, num_minutes, num_seconds):
    """
    Computes the total number of seconds from days, hours, minutes and seconds.

    Args:
    All the inputs must be non-negative.
    num_days: number of days.
    num_hours: number of hours.
    num_minutes: number of minutes.
    num_seconds: number of seconds.

    Returns:
    Integer containing the total number of seconds.

    Raises:
    An AssertionError if any of the inputs is negative.
    """
    # AssertionError if any of the inputs is negative.
    assert num_days >= 0 and num_hours >= 0 and num_minutes >= 0 and num_seconds >= 0, "All inputs must be non-negative"

    return num_days * 24 * 60 * 60 + num_hours * 60 * 60 + num_minutes * 60 + num_seconds
```

```python
compute_seconds(1,2,3,4)
```

```python
compute_seconds(1,2,3,-4)
```
