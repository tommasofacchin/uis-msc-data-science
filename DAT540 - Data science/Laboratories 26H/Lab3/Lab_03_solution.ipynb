### Question 8

Create a function to
- generate a 2D square matrix of 3 elements
- make the diagonal values equals to 10
- perform a matrix dot product on the changed matrix

*Note*: 2D numpy array is analogous to matrix in the world of data science.

```python
def square_matrix(x):
  # generate a 2D square matrix of 3 elements
  matrix = np.arange(x*x).reshape(x, x)

  # make the diagonal values equals to 10
  np.fill_diagonal(matrix, 10)

  # perform a matrix dot product on the changed matrix
  return np.dot(matrix, matrix.T)


print(square_matrix(3))
```

### Question 9

Given a dataset of exam scores for a class of students,
- Calculate the mean (average) score of the class.
- Calculate the median score of the class.
- Calculate the minimum and maximum scores in the dataset.
- Calculate the range of scores (the difference between the maximum and minimum scores).
- Calculate the variance of the scores.
- Calculate the standard deviation of the scores.

Provide the results for each of these statistical operations.

```python
scores = np.array([85, 92, 88, 78, 95, 89, 92, 87, 91, 84])
```

```python
# Calculate the mean (average) score of the class.
mean_score = np.mean(scores)

# Calculate the median score of the class.
median_score = np.median(scores)

# Calculate the minimum and maximum scores in the dataset.
min_score = np.min(scores)
max_score = np.max(scores)

# Calculate the range of scores (the difference between the maximum and minimum scores).
score_range = max_score - min_score

# Calculate the variance of the scores.
variance_score = np.var(scores)

# Calculate the standard deviation of the scores.
std_deviation_score = np.std(scores)

# Print the results
print("Mean score:", mean_score)
print("Median score:", median_score)
print("Minimum score:", min_score)
print("Maximum score:", max_score)
print("Range of scores:", score_range)
print("Variance of scores:", variance_score)
print("Standard deviation of scores:", std_deviation_score)
```
