### Sorting Arrays
**1.** Print the three smallest elements of the defined numpy array in ascending order.

````python
array1 = np.array([10, -4, 0, 23, 4, 2, 2])
````

````python
# Solution

# Find the indices that would sort the array
sorted_indices = np.argsort(array1)

# Extract the three smallest elements using the sorted indices
three_smallest = array1[sorted_indices[:3]]

# Print the three smallest elements in ascending order
print(three_smallest)
````

````python
# Alternative solution
np.sort(array1)[0:3]
````

**2.** Print the array in descending order

````python
# Solution

# Sort the array in descending order
sorted_arr = np.sort(array1)[::-1]

# Print the sorted array
print(sorted_arr)
````

````python
# Alternative solution
array1.sort()  # sort array in place
array1[::-1]
````

**8.** Divide your random matrix by the vector ``[1,2,2,2,3]`` such that each item of the vector divides the respective row of your random matrix and print the resultant array.

Hint: You can use broadcasting to achieve this. Broadcasting in NumPy is the ability to perform element-wise operations on arrays with different shapes, where NumPy automatically adjusts the shapes of the arrays to make the operation possible.

````python
# Solution

division_vector = np.array([1, 2, 2, 2, 3])

# Divide each row of the matrix by the division vector. 
# In this solution we transpose the random matrix so that the division happens row wise. 
# Then we have to transpose the resultant array again to show our initial columns/rows.
result_array = (random_array.T / division_vector).T

# Print the resultant array
print(result_array)

# Solution 2 (Uses broadcasting)
print(random_array / division_vector[:, None])
````

**17.** Calculate Gini Impurity

You are working on building a decision tree for a classification problem. Gini impurity is a measure of the impurity or disorder of a set of data points. For a binary classification problem, Gini impurity for a node with data points is calculated as follows:

$$
\text{Gini Impurity} = 1 - \sum_{i=1}^{k} (p_i)^2
$$

Where:
- \(k\) is the number of classes (in this binary case, \(k = 2\)).
- \(p_i\) is the proportion of data points in class \(i\) among the data points at the node.

Your task is to write a Python function that calculates the Gini impurity for a given set of data points.

You should create a function named `calculate_gini_impurity(data)` where `data` is a NumPy array containing the class labels for a set of data points. The function should return the Gini impurity.

For example, if you have the following data:

```python
data = np.array([0, 1, 0, 1, 1, 0])
```

You should be able to call your function like this:

```python
gini_impurity = calculate_gini_impurity(data)
print(gini_impurity)
```

And it should output the Gini impurity value for the given data.

Please write the Python function to calculate Gini impurity as described above.

````python
data = np.array([0, 1, 0, 1, 1, 0])
````

````python
# Solution

def calculate_gini_impurity(data):
    # Count the occurrences of each class label in the data
    class_counts = np.bincount(data)
    
    # Calculate the proportion of each class
    class_probs = class_counts / len(data)
    
    # Calculate Gini impurity
    gini_impurity = 1 - np.sum(class_probs**2)
    
    return gini_impurity

# Example usage:
gini_impurity = calculate_gini_impurity(data)
print("Gini Impurity:", gini_impurity)
````
