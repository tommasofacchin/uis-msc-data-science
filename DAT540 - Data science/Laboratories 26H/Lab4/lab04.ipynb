**17.** Calculate Gini Impurity

You are working on building a decision tree for a classification problem. Gini impurity is a measure of the impurity or disorder of a set of data points. For a binary classification problem, Gini impurity for a node with data points is calculated as follows:

$$
\text{Gini Impurity} = 1 - \sum_{i=1}^{k} (p_i)^2
$$

Where:
- \(k\) is the number of classes (in this binary case, \(k = 2\)).
- \(p_i\) is the proportion of data points in class \(i\) among the data points at the node.

Your task is to write a Python function that calculates the Gini impurity for a given set of data points.
