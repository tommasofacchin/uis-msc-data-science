### Question 2 (cleaning the dataframe)
1. Check if any null values in each column of the dataframe (*Hint*: use function .isnull() and .sum() )
2. Create a new dataframe "cat_no_null" without null values i.e drop all rows with any null value (*Hint*: use function .dropna(how=) )
3. Create a new dataframe "cat" by replacing null values by value "other" (*Hint*: use function .fillna(value= ) )
4. Check if any row are duplicated in the dataframe (*Hint*: use function .duplicated() )
5. Drop "inplace" the duplicated rows in the same dataframe "cat" (*Hint*: use function .drop_duplicates(inplace= ) )

### Question 10 (dataframe "cat" and computation with Categoricals)
1. Create dataframe "cat_w" that select only the column "weight"
2. Create a pandas series "quartile" that discretizes the variable "weight" into 4 equal-sized buckets (Q1, Q2, Q3, Q4). (***Hint***: use functions pd.qcut() and then pd.Series() )
3. Create a dataframe that display the 4 quartiles with columns (count, min, max)

```python
# ´Solution

# Question 10.1
cat_w = cat.loc[:, "weight"]

# Question 10.2
bins = pd.qcut(cat_w, 4, labels=['Q1', 'Q2', 'Q3', 'Q4'])
bins
```

```python
# Question 10.3

bins = pd.Series(bins, name='quartile')
results = (pd.Series(cat_w)
          .groupby(bins)
          .agg(['count', 'min', 'max'])
          .reset_index())
results
```
