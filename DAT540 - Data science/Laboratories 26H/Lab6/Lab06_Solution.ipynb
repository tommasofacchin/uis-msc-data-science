### Question 05
**a)** Given a dataset containing information about cars, including their makes and colors. Your task is to create a one-hot encoded dataframe that represents the car makes and colors as binary features. 
<br>**b)** Given a dataset containing information about students and their courses. Your task is to create a DataFrame that shows the total number of subjects taken by each student, with one-hot encoding for the course subjects. Each student can take multiple courses, and courses can belong to different subjects. 
<br> **c)** Given a dataset containing information about citizens of different cities. Find the citizens whose age is greater than 30. Show this in a dataframe in a dataframe in descending order with changing the name of the column 'Name' to 'First Name' and finally, setting 'First Name' as the index for this dataframe. You can perform all these changes and filtering by chaining the methods together.

```python
# a)
# Sample input data
data = {
    'Car': ['Car_1', 'Car_2', 'Car_3', 'Car_4', 'Car_5'],
    'Make': ['Ford', 'Toyota', 'Ford', 'Toyota', 'Honda'],
    'Color': ['Red', 'Blue', 'Green', 'Red', 'Blue']
}

cars = pd.DataFrame(data)
dummies_cars = pd.get_dummies(cars[['Make', 'Color']])
encoded_cars = pd.concat([cars, dummies_cars], axis=1)
encoded_cars.drop(['Make', 'Color'], axis=1, inplace=True)
encoded_cars
```

```python
# b)
# Sample input data
data = {
    'Student_ID': [1001, 1001, 1002, 1003, 1003, 1003],
    'Course_ID': ['C101', 'C102', 'C103', 'C104', 'C105', 'C106'],
    'Subject': ['Math', 'Science', 'English', 'Math', 'History', 'Science']
}
students = pd.DataFrame(data)

dummies_students = pd.get_dummies(students['Subject'])
students = pd.concat([students, dummies_students], axis=1)
grouped_students = students.groupby('Student_ID').sum()
grouped_students.reset_index(inplace=True)
grouped_students
```

```python
# c)
# Sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, 35, 40, 45],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Miami']
}

citizen = pd.DataFrame(data)

result = (
    citizen
    .loc[citizen['Age'] > 30]            # Filter rows where Age > 30
    .sort_values(by='Age', ascending=False)  # Sort by Age in descending order
    .rename(columns={'Name': 'First Name'})  # Rename the 'Name' column
    .set_index('First Name')         # Set 'First Name' as the index
)
result
```

### Question 10
Using the dataset from Question 01,

<br>**a)** Label the age, 20-29 == 'Under 30', 29 - 40 == 'Adult', 40 - 60 == 'Middle Aged' and 60 - 110 == 'Senior', and create a column named 'age_group' to store these values.

<br>**b)** Select Customers in different age group who is Married i.e Marital_Status == Married. Find their on average total revolving balance for each category of dependent count.

<br>**c)** Select Customers in different age group who is Married i.e Marital_Status == Married. Create a contingency table with count for each category of dependent count indexed by the age group. Additionally, drop n/a values.

Hint Useful functions: pd.replace, pd.cut, df.query(),
        

```python
# a)
def label_age(df):
    bins = [20, 29, 40, 60, 110]
    labels = ['Under 30', 'Adult', 'Middle Aged', 'Senior']
    age_group = pd.cut(df['Customer_Age'], labels = labels, bins = bins)
    df['age_group'] = age_group
    return df

label_age(combined_customer_data)
```

```python
# b) Pivot Table
df_a = (combined_customer_data
    .pipe(label_age)
    .query('Marital_Status == "Married"')
    .dropna()
    .pivot_table(
        values="Total_Revolving_Bal",
        columns="Dependent_count",
        index="age_group",
        aggfunc='mean'
    ))

df_a
```

```python
# c)  ContingencyTable
df_b=(combined_customer_data
    .pipe(label_age)
    .query('Marital_Status == "Married"')
    .dropna()
    .pipe(lambda x: pd.crosstab(index=x["age_group"], columns=x["Dependent_count"])))
df_b
```
