### Question 05
In order to perform numerical calculations on data, it is usual to encode string data into numerical formats.<br>
For categorical data this is achieved by applying one-hot encoding or sparse encoding. The latter is easily achieved using the map function.
<br>**a)** Create a copy of the `df` dataframe called `df_num`
<br>**b)** Convert values in each `object` feature into integers using the mapping rule:

* gender: {'male': 0, 'female': 1}
* customer type: {'disloyal customer': 0, 'loyal customer': 1}
* type of travel: {'personal travel': 0, 'business travel': 1}
* class: {'eco': 0, 'eco plus': 1, 'business': 2}
* satisfaction: {'satisfied': 1, 'neutral or dissatisfied': 0}

<br>**c)** For each class display the percentage of passengers that expressed overall satisfaction.

```python
df_num = df.copy()
```

```python
if df_num['gender'].dtype!='int': # for sanity check only
    df_num['gender'] = df_num['gender'].map({'male':0, 'female':1})
    df_num['customer type'] = df_num['customer type'].map({'disloyal customer': 0, 'loyal customer': 1})
    df_num['type of travel'] =  df_num['type of travel'].map({'personal travel': 0, 'business travel': 1})
    df_num['class'] = df_num['class'].map({'eco': 0, 'eco plus': 1, 'business': 2})
    df_num['satisfaction'] = df_num['satisfaction'].map({'satisfied': 1, 'neutral or dissatisfied': 0})
```

```python
df_num.groupby(by=['class'])[['satisfaction']].agg(lambda x: 100*sum(x==1)/len(x)).round(2)
```

### Question 06
A very useful numerical data transformation in ML is known as _normalization_.<br>
For the dataframe `df_num` we shall normalize each columnn using the MinMax transformation: 
$ z \to \dfrac{x-x_{\min}}{x_{\max} - x_{\min}}.$
<br>**a)** Derive two pandas series `df_min` and `df_max` comprised of the mininum and maximum values of each column of `df_num` respectively.
<br>**b)** Apply the MinMax normalization to each column of `df_num`. Name the transformed dataframe `df_norm`.
<br>**c)** Explain what this transformation does to each column.
<br>**d)** State columns whose values remained unchanged under this transformation. What change occurs in this actually happens for these column?

```python
df_min = df_num.min()
df_max = df_num.max()
```

```python
df_norm = (df_num - df_min) / (df_max - df_min)
```

* This MinMax transformation applies a linear scaling of the values of each column into float type values in the range 0.0 to 1.0.

* Columns with binary values (0 and 1) represented by categorical features (gender, customer type, travel type, satisfaction) remain unchanged under this transformation.  The only thing that changes is their dtype (from `int` for `float`).

Reverse category mapping:<br>
<br>**a)** Write a function that takes a list/series of integers representing satisfacion values and returns a corresponding list of categorical labels (strings).
<br>**b)** Verify that your function in **a)** works as expected, by using a random sample of 10 satisfaction labels drawn from `df_test`.
<br>**c)** Write a function that takes a list/series integers representing passenger class values and returns a corresponding list of passenger classes (strings).
<br>**d)** Verify that your function in **c)** works as expected, by using a random sample of 10 passenger class values drawn from `df_test`.

```python
from numpy import ndarray

def inv_mapping0(sample):
    if isinstance(sample, (list, ndarray)):
        sample = pd.Series(sample)  # convert input to pandas series
    return sample.map({0:'neutral or dissatisfied', 1:'satisfied'})
```

```python
df_test_sample = pd.DataFrame(columns=['satisfaction'])
df_test_sample['satisfaction'] = df_test['satisfaction'].sample(10)
df_test_sample['category'] = inv_mapping0(df_test_sample['satisfaction'])
df_test_sample
```

```python
def inv_mapping1(sample):
    if isinstance(sample, (list, ndarray)):
        sample = pd.Series(sample)  # convert input to pandas series
    sample = (sample * 2).astype('int')
    return sample.map({0:'eco', 1:'eco plus', 2:'business'})
```
